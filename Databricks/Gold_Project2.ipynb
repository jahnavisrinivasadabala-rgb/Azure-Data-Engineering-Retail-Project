```python
from pyspark.sql.functions import col

gold_orders_df = spark.table("retail_catalog.silver.orders")

gold_orders_df.show()

gold_order_items_df=spark.table("retail_catalog.silver.order_items")

gold_order_items_df.show()

fact_sales_df = gold_orders_df.join(gold_order_items_df ,"order_id","inner").filter(col("order_status")=="COMPLETED").withColumn("sales_amount",col("quantity") * col("selling_price"))

silver_products_df = spark.table(
    "retail_catalog.silver.products"
)
dim_product_df = silver_products_df.select(
    "product_id",
    "product_name",
    "category",
    "brand",
    "unit_price",
    "supplier"
)

silver_customers_df = spark.table(
    "retail_catalog.silver.customers"
)
dim_customers_df = silver_customers_df.select ("customer_id",
"first_name",
"last_name",
"gender",
"email",
"phone",
"city",
"state",
"country",
"signup_date")
dim_customers_df.printSchema()

silver_stores_df = spark.table(
    "retail_catalog.silver.stores"
)

dim_stores_df = silver_stores_df.select("store_id",
"store_name",
"city",
"state",
"country",
"store_address",
"phone")
```
